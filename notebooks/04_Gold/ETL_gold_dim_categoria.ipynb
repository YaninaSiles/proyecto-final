```sql
%sql

INSERT OVERWRITE proyecto_final.gold.dim_categoria(
  tipo_entidad,
  subcategoria
)
WITH datos_nuevos AS (
    SELECT DISTINCT tipo_entidad, tipo_escuela AS subcategoria FROM proyecto_final.silver.escuelas
    UNION 
    SELECT DISTINCT tipo_entidad, especialidad as subcategoria FROM proyecto_final.silver.hospitales
    UNION 
    SELECT DISTINCT tipo_entidad, categoria as subcategoria FROM proyecto_final.silver.kioscos
    UNION 
    SELECT DISTINCT tipo_entidad, unidad_academica as subcategoria FROM proyecto_final.silver.universidades
)
SELECT * FROM datos_nuevos;
```
