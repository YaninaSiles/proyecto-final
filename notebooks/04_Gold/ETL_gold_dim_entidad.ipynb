```sql
%sql

INSERT OVERWRITE proyecto_final.gold.dim_entidad(
  nombre_entidad,
  direccion,
  telefono,
  sitio_web,
  coordenadas
)
WITH datos AS (
  SELECT nombre_establecimiento AS nombre_entidad, direccion, CAST(NULL AS STRING) AS telefono, CAST(NULL AS STRING) AS sitio_web, coordenadas FROM proyecto_final.silver.escuelas
  UNION ALL
  SELECT nombre_hospital, direccion, telefono, sitio_web, coordenadas FROM proyecto_final.silver.hospitales
  UNION ALL
  SELECT nombre, direccion, telefono, sitio_web, coordenadas FROM proyecto_final.silver.kioscos
  UNION ALL
  SELECT nombre_universidad, direccion, telefono, sitio_web, coordenadas FROM proyecto_final.silver.universidades
)
SELECT * from datos;
```
