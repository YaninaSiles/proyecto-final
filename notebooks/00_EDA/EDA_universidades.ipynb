***EDA Universidades***

Se filtrará por valores unicos Nombre - Dirreccion - geometry

Valores nulos asignar otro valor como 'No tiene'.

Las coordenadas en geometry ya estan en grados decimales válidos, solo debo separar en longitud y latitud.



```sql
DESCRIBE proyecto_final.raw.universidades_bronze

--unidad_aca --> unidad academica
--direreccion --> direccion
```

```sql
%sql 
select *
from proyecto_final.raw.universidades_bronze
limit 200;
```

```sql
%sql

--Nulos

select 
count(*) as total,
count(*) - count(id) as id_nulos,
count(*) - count(nombre) as nombre_nulos,
count(*) - count(unidad_aca) as unidad_aca_nulos,
count(*) - count(sede) as sede_nulos,
count(*) - count(dirreccion) as dirreccion_nulos,
count(*) - count(barrio) as barrio_nulos,
count(*) - count(comuna) as comuna_nulos,
count(*) - count(telefono) as telefono_nulos,
count(*) - count(fax) as fax_nulos,
count(*) - count(web) as web_nulos,
count(*) - count(geometry) as geometry_nulos,
count(*) - count(_rescued_data) as _rescued_data_nulos
from proyecto_final.raw.universidades_bronze

-- En campos importantes no hay nulos: id nombre comuna geometry
```

```sql
%sql

-- Duplicados
select nombre, dirreccion,geometry, COUNT(*) as duplicados
from proyecto_final.raw.universidades_bronze
group by nombre, dirreccion, geometry
having COUNT(*) > 1;
```

```sql
%sql
select barrio, count(*) as cantidad
from proyecto_final.raw.universidades_bronze
group by barrio
order by cantidad desc

-- 24 barrios 
-- Hay mas en Recoleta - Balvanera - San Nicolas - Monserrat
```

```sql
%sql

--13 comunas. Comuna 1 con mas universidades

-- En escuelas el campo comuna son numeros, puedo aca sacar Comuna aca, o agregarle en las escuelas 
select comuna, count(*) as cantidad
from proyecto_final.raw.universidades_bronze
group by comuna
order by cantidad desc
```

```sql
%sql
-- Coordenadas dentro de CABA

with coordenadas as (
  SELECT geometry coordenadas, -- long lat,
  st_x(st_geomfromtext(geometry)) AS longitud,
  st_y(st_geomfromtext(geometry)) AS latitud
  FROM proyecto_final.raw.universidades_bronze
)
select * from coordenadas
WHERE latitud > -34.50 OR latitud < -34.70 
   OR longitud > -58.30 OR longitud < -58.55;

```
