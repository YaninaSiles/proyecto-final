```sql
%sql

DROP TABLE IF EXISTS proyecto_final.silver.barrios;


CREATE TABLE IF NOT EXISTS proyecto_final.silver.barrios (
  barrio_id INT COMMENT 'ID único del barrio',
  barrio_nombre STRING COMMENT 'Nombre del barrio',
  tipo_entidad STRING COMMENT 'Tipo de entidad',
  comuna_id INT COMMENT 'ID de la comuna donde se encuentra el barrio',
  perimetro_m DOUBLE COMMENT 'Périmetro en metros del barrio',
  area_m2 DOUBLE COMMENT 'Área en metros cuadrados del barrio',
  coordenadas STRING COMMENT 'Coordenadas del barrio de tipo POLYGON que indica la ubicación del barrio ',
  _source_table STRING DEFAULT 'proyecto_final.raw.barrios_bronze' COMMENT 'Tabla de origen',
  _processing_time TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
) USING DELTA
TBLPROPERTIES(
  'delta.feature.allowColumnDefaults' = 'supported'
)
COMMENT 'Barrios de la Ciudad Autónoma de Buenos Aires';
```
