```sql
%sql

DROP TABLE IF EXISTS proyecto_final.silver.escuelas;

CREATE TABLE IF NOT EXISTS proyecto_final.silver.escuelas (
  id_establecimiento INT COMMENT 'ID de la escuela',
  nombre_funcional STRING COMMENT 'Nombre completo',
  nombre_generico STRING COMMENT 'Tipo de establecimiento',
  nombre_establecimiento STRING COMMENT 'Nombre de la escuela',
  codigo_cui INT COMMENT 'Código Único de Infraestructura ',
  codigo_cue INT COMMENT 'Código Único de Establecimiento',
  codigo_anexo INT COMMENT 'Ambos códigos unidos',
  tipo_gestion STRING COMMENT 'Tipo de escuela: estatal o privada',
  nivel_educativo STRING COMMENT 'Tipo de nivel educativo',
  tipo_escuela STRING COMMENT 'Tipo de escuela',
  dependencia STRING COMMENT 'Tipo de Direccion',
  direccion STRING,
  barrio STRING,
  comuna INT,
  fuente_datos STRING COMMENT 'Origen de los datos',
  coordenadas STRING COMMENT 'Coordenadas donde se ubica la escuela',
  tipo_entidad STRING COMMENT 'Tipo de entidad: Escuela',
  longitud DOUBLE COMMENT 'Longitud en grados decimales',
  latitud DOUBLE COMMENT 'Latitud en grados decimales',
  _processing_time TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
) USING DELTA
TBLPROPERTIES(
  'delta.feature.allowColumnDefaults' = 'supported'
)
COMMENT 'Escuelas iniciales, primarias y secundarias de la Ciudad Autónoma de Buenos Aires';
```
