```sql
%sql

DROP TABLE IF EXISTS proyecto_final.gold.dim_comunas;

CREATE TABLE IF NOT EXISTS proyecto_final.gold.dim_comunas (
  id BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1) COMMENT 'PK de la dimensión',
  nro_comuna INT COMMENT 'Número de la comuna',
  perimetro_m DOUBLE COMMENT 'Perimetro en metros',
  area_m2 DOUBLE COMMENT 'Área en metros cuadrados',
  coordenadas STRING COMMENT 'Polígono de la comuna',
  _created_at TIMESTAMP DEFAULT current_timestamp()
) USING DELTA
TBLPROPERTIES (
  'delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Dimensión de comunas de CABA';
```
