```sql
%sql

DROP TABLE IF EXISTS proyecto_final.gold.dim_entidad;

CREATE TABLE IF NOT EXISTS proyecto_final.gold.dim_entidad (
  id BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1) COMMENT 'PK de la dimensión',
  nombre_entidad STRING COMMENT 'Nombre comercial o institucional',
  direccion STRING COMMENT 'Dirección física',
  telefono STRING,
  sitio_web STRING,
  latitud DOUBLE COMMENT 'Latitud exacta del punto',
  longitud DOUBLE COMMENT 'Longitud exacta del punto',
  coordenadas STRING COMMENT 'Punto WKT (POINT) para mapas',
  _created_at TIMESTAMP DEFAULT current_timestamp()
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Dimensión con los datos de contacto y ubicación exacta de la entidad';
```
