```sql
%sql

DROP TABLE IF EXISTS proyecto_final.silver.comunas;

CREATE TABLE IF NOT EXISTS proyecto_final.silver.comunas (
  comuna_id INT COMMENT 'ID de la comuna',
  tipo_entidad STRING COMMENT 'Tipo de entidad',
  barrios STRING COMMENT 'Barrios dentro de la comuna',
  perimetro_m DOUBLE COMMENT 'Périmetro en metros de la comuna',
  area_m2 DOUBLE COMMENT 'Área en metros cuadrados de la comuna',
  coordenadas STRING COMMENT 'Coordenadas de la comuna de tipo POLYGON ',
  _processing_time TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
) USING DELTA
TBLPROPERTIES(
  'delta.feature.allowColumnDefaults' = 'supported'
)
COMMENT 'Comunas de la Ciudad Autónoma de Buenos Aires';
```
