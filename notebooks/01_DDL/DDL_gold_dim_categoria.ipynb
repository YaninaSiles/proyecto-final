```sql
%sql

DROP TABLE IF EXISTS proyecto_final.gold.dim_categoria;

CREATE TABLE IF NOT EXISTS proyecto_final.gold.dim_categoria (
  id BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1) COMMENT 'PK de la categoria',
  tipo_entidad STRING COMMENT 'Kiosco, Escuela, Hospital o Universidad',
  subcategoria STRING COMMENT 'Nivel educativo, especialidad u otro detalle',
  _created_at TIMESTAMP DEFAULT current_timestamp()
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Dimensión para clasificar y filtrar los puntos de interés';
```
