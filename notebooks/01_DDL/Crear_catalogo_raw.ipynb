```sql
%sql
-- CREO MI CATÁLAGO

CREATE CATALOG IF NOT EXISTS proyecto_final;

SHOW CATALOGS;
```

```sql
%sql

-- CREO EL ESQUEMA RAW

USE CATALOG proyecto_final;

CREATE SCHEMA IF NOT EXISTS proyecto_final.raw;

SHOW SCHEMAS;
```

```sql
%sql

-- CREO VOLUME

CREATE VOLUME IF NOT EXISTS proyecto_final.raw.archivos
COMMENT 'Volume para almacenar archivos CSV crudos';
    
SHOW VOLUMES IN proyecto_final.raw;
    

-- CREO VOLUME
```
