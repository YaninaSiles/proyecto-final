```sql
%sql

DROP TABLE IF EXISTS proyecto_final.silver.kioscos;

CREATE TABLE IF NOT EXISTS proyecto_final.silver.kioscos (
  nombre STRING COMMENT 'Nombre del kiosco',
  barrio STRING COMMENT 'Barrio obtenido con las coordenadas del kiosco buscado en el dataset de barrios',
  comuna INT COMMENT 'Comuna obtenido con las coordenadas del kiosco buscado en el dataset de barrios',
  categoria STRING COMMENT 'Categoria del kiosco',
  direccion STRING,
  telefono STRING,
  sitio_web STRING,
  email STRING,
  instagram STRING,
  facebook STRING,
  rating STRING COMMENT 'Puntaje',
  reviews_count STRING COMMENT 'Cantidad de reviews',
  claimed BOOLEAN COMMENT 'Indica si el kiosco fue validado por el dueño',
  top_review STRING COMMENT 'Review mas alta',
  horario STRING COMMENT 'Horario de atención',
  latitud DOUBLE COMMENT 'Latitud en grados decimales',
  longitud DOUBLE COMMENT 'Longitud en grados decimales',
  tipo_entidad STRING COMMENT 'Tipo entidad: Kiosco',
  coordenadas STRING COMMENT 'Coordenadas donde se encuentra el kiosco',
  _processing_time TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
) USING DELTA
TBLPROPERTIES(
  'delta.feature.allowColumnDefaults' = 'supported'
)
COMMENT 'Kioscos de la Ciudad Autónoma de Buenos Aires';
```
