```sql
%sql

DROP TABLE IF EXISTS proyecto_final.gold.fact_entidades;

CREATE TABLE IF NOT EXISTS proyecto_final.gold.fact_entidades (
  hecho_id BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1) COMMENT 'PK del hecho',
  
  -- fk
  entidad_id BIGINT COMMENT 'FK a dim_entidad',
  categoria_id BIGINT COMMENT 'FK a dim_categoria',
  barrio_id BIGINT COMMENT 'FK a dim_barrios',
  comuna_id BIGINT COMMENT 'FK a dim_comunas',

  latitud DOUBLE COMMENT 'Latitud de la entidad',
  longitud DOUBLE COMMENT 'Longitud de la entidad',

  --metricas
  es_generador_trafico INT COMMENT 'Vale 1 si es Hospital, Escuela o Univ. Vale 0 si es Kiosco',
  es_competencia INT COMMENT 'Vale 1 si es Kiosco. Vale 0 en el resto',

  _created_at TIMESTAMP DEFAULT current_timestamp()
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Tabla de hechos para analizar densidad de competencia vs generación de tráfico en CABA';
```
