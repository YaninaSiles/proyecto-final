```sql
%sql
DROP TABLE IF EXISTS proyecto_final.raw.escuelas_bronze;

CREATE TABLE proyecto_final.raw.escuelas_bronze
SELECT 
id as id_establecimiento,
    fna as nombre_funcional,
    gna as nombre_generico,
    nam as nombre_establecimiento,
    cui as codigo_cui,
    cue as codigo_cue,
    anx as codigo_anexo,
    ges as tipo_gestion,
    nen_mde as nivel_educativo,
    tip as tipo_escuela,
    dep as dependencia, 
    dir as direccion,
    bar as barrio,
    com as comuna,
    sag as fuente_datos, 
    geometry as geometria_wkt 
FROM read_files(
  '/Volumes/proyecto_final/raw/archivos/establecimientos_educativos.csv',
  format => 'csv',
  header => true
);
```
