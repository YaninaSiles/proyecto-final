```sql
%sql

DROP TABLE IF EXISTS proyecto_final.silver.universidades;

CREATE TABLE IF NOT EXISTS proyecto_final.silver.universidades (
  id_universidad INT COMMENT 'ID de la universidad proveniente de origen',
  nombre_universidad STRING COMMENT 'Nombre de la universidad completa',
  unidad_academica STRING,
  sede STRING,
  direccion STRING,
  barrio STRING,
  comuna INT,
  telefono STRING,
  fax STRING,
  sitio_web STRING,
  longitud DOUBLE COMMENT 'Longitud en grados decimales',
  latitud DOUBLE COMMENT 'Latitud en grados decimales',
  coordenadas STRING COMMENT 'Coordenadas de donde se ubica la universidad',
  tipo_entidad STRING COMMENT 'Tipo entidad: Universidad',
  _processing_time TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
) USING DELTA
TBLPROPERTIES(
  'delta.feature.allowColumnDefaults' = 'supported'
)
COMMENT 'Universidades de la Ciudad Autónoma de Buenos Aires';
```
