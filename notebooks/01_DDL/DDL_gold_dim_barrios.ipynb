```sql
%sql

DROP TABLE IF EXISTS proyecto_final.gold.dim_barrios;

CREATE TABLE IF NOT EXISTS proyecto_final.gold.dim_barrios (
  id BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1) COMMENT 'PK de la dimensión',
  barrio_id_origen INT COMMENT '',
  barrio_nombre STRING COMMENT '',
  perimetro_m DOUBLE COMMENT '',
  area_m2 DOUBLE COMMENT '',
  coordenadas STRING COMMENT '',
  _created_at TIMESTAMP DEFAULT current_timestamp()
) USING DELTA
TBLPROPERTIES (
  'delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Dimensión de barrios de CABA';
```
