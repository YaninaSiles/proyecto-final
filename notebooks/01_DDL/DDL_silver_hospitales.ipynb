```sql
%sql

DROP TABLE IF EXISTS proyecto_final.silver.hospitales;

CREATE TABLE IF NOT EXISTS proyecto_final.silver.hospitales (
  id_hospital BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1) COMMENT 'ID del hospital generado', 
  nombre_hospital STRING COMMENT 'Nombre del hospital',
  nombre_funcional_completo STRING COMMENT 'Nombre completo',
  nombre_generico STRING COMMENT 'Nombre genérico',
  especialidad STRING,
  tipo_atencion STRING,
  direccion STRING ,
  barrio STRING ,
  fuente_datos STRING COMMENT 'Origen de los datos',
  comuna INT ,
  telefono STRING ,
  sitio_web STRING,
  coordenadas STRING COMMENT 'Coordenadas donde se encuentra el hospital',
  tipo_entidad STRING COMMENT 'Tipo de entidad: Hospital',
  longitud DOUBLE COMMENT 'Longitud en grados decimales',
  latitud DOUBLE COMMENT 'Latitud en grados decimales',
  _processing_time TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
) USING DELTA
TBLPROPERTIES(
  'delta.feature.allowColumnDefaults' = 'supported'
)
COMMENT 'Hospitales de la Ciudad Autónoma de Buenos Aires';
```
